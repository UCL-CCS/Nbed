```python
from nbed import NbedConfig
from nbed.driver import NbedDriver

complex = 3
xyz = inputs[complex]["xyz"]
for charge in inputs[complex]["charge"]:
    for multiplicity in inputs[complex]["multiplicity"]:
        # multiplicity=inputs[complex]["multiplicity"][0]
        spin = multiplicity - 1

        print(f"Complex:{complex}, charge:{charge}, spin:{spin}")

        config = NbedConfig(
            geometry=xyz,
            n_active_atoms=1,
            basis="sto-3g",
            xc_functional="PBE",
            charge=charge,
            spin=spin,
            run_ccsd_emb=False,
            projector="both",
            init_huzinaga_rhf_with_mu=True,
            localization="spade",
            virtual_localization="cl",
            build_hamiltonian=False,
            run_dft_in_dft=True,
            restricted_global=False,
            restricted_active=False,
            max_hf_cycles=200,
            max_dft_cycles=200,
        )

        driver = NbedDriver(config)
        driver.embed()
        mu_mol = driver.mu["scf"].mol
        print()
        print("RESULTS:")
        print(mu_mol.spin, mu_mol.charge)
        break
```

```python
import numpy as np

from nbed.driver import run_emb_ccsd

nmos = driver.mu["cl"].shells
print(nmos)
n_occ = np.sum(
    np.bitwise_or(
        driver.localized_system.active_occ_inds[0],
        driver.localized_system.active_occ_inds[1],
    )
)


def partial_ccsd(shell_index):
    """CCSD with restriction."""
    print("Freezing orbitals above %s", shell_index)
    scf = driver.mu["scf"].copy()
    frozen = [*range(shell_index, nmos)]
    ccsd, _ = run_emb_ccsd(scf, frozen=frozen)
    energy = ccsd.e_tot - driver.e_nuc + driver.mu["classical_energy"]
    return energy
```

```python
ccsd_energies = {index: partial_ccsd(index) for index in driver.mu["cl"].shells}
```

```python
ccsd_energies = {int(k): v for k, v in ccsd_energies.items()}
ccsd_energies = dict(sorted(ccsd_energies.items()))
```
