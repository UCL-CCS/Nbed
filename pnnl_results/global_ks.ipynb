```python
import numpy as np

np.array([0]).shape[-1]
```

```python
from pyscf import gto, scf

conv_tol = 1e-6
results = []
for complex, input in inputs.items():
    geom = input["xyz"]
    for charge in input["charge"]:
        for multiplicity in input["multiplicity"]:
            print(
                f"Geom={geom}, Charge={charge}, Multiplicity={multiplicity}, ConvTol={conv_tol}"
            )
            mol = gto.Mole(
                atom=geom[2:], basis="sto-3g", charge=charge, multiplicity=multiplicity
            ).build()
            mf = scf.UKS(mol)
            mf.xc = "PBE"
            mf.max_cycle = 1000
            mf.diis_space = 12
            mf.conv_tol = conv_tol
            mfresult = mf.kernel()
            this_result = {
                "complex": complex,
                "charge": charge,
                "multiplicity": multiplicity,
                "XC": "PBE96",
                "energy": mfresult,
                "spin": mf.spin_square(),
            }
            results.append(this_result)
            print()
```
